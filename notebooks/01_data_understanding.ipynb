```python
print("Year statistics:")
print(df['year'].describe())

print("\nSelling price statistics:")
print(df['selling_price'].describe())

print("\nKilometers driven statistics:")
print(df['km_driven'].describe())
```

```python
print("Cars before 1990:")
print((df['year'] < 1990).sum())

print("\nCars after 2020:")
print((df['year'] > 2020).sum())

print("\nZero/negative selling prices:")
print((df['selling_price'] <= 0).sum())

print("\nZero/negative kilometers:")
print((df['km_driven'] <= 0).sum())
```
