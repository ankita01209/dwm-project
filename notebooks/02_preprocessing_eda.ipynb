```python
brand_avg_price = (
    df.groupby('brand')['selling_price']
      .mean()
      .sort_values(ascending=False)
)

print(brand_avg_price)
```

```python
plt.figure(figsize=(12, 8))

brand_avg_price.plot(kind='bar')

plt.title('Average Selling Price by Car Brand')
plt.xlabel('Brand')
plt.ylabel('Average Selling Price (₹)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
brand_median_price = (
    df.groupby('brand')['selling_price']
      .median()
      .sort_values(ascending=False)
)

print(brand_median_price)
```

Since selling price is strongly right-skewed, the median provides a more robust measure of the typical price than the mean. The brand-level median prices show substantial differences between brands, with premium brands generally having higher typical prices than mass-market brands.

```python
fuel_price_summary = df.groupby('fuel')['selling_price'].agg(
    ['count', 'mean', 'median']
).sort_values('median', ascending=False)

print(fuel_price_summary)
```

```python
transmission_price_summary = df.groupby('transmission')['selling_price'].agg(
    ['count', 'mean', 'median']
).sort_values('median', ascending=False)

print(transmission_price_summary)
```

```python
owner_price_summary = df.groupby('owner')['selling_price'].agg(
    ['count', 'mean', 'median']
).sort_values('median', ascending=False)

print(owner_price_summary)
```

The analysis shows a general decrease in observed selling price as the number of previous owners increases. First-owner vehicles have a median selling price of approximately ₹4.50 lakh, while fourth-or-above-owner vehicles have a median of ₹1.38 lakh. Test-drive cars show a high median price, but this category contains only 17 observations, so the result should be interpreted cautiously.

```python
Q1 = df['selling_price'].quantile(0.25)
Q3 = df['selling_price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

price_outliers = df[
    (df['selling_price'] < lower_bound) |
    (df['selling_price'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of price outliers:", len(price_outliers))
```

```python
Q1_km = df['km_driven'].quantile(0.25)
Q3_km = df['km_driven'].quantile(0.75)

IQR_km = Q3_km - Q1_km

lower_bound_km = Q1_km - 1.5 * IQR_km
upper_bound_km = Q3_km + 1.5 * IQR_km

km_outliers = df[
    (df['km_driven'] < lower_bound_km) |
    (df['km_driven'] > upper_bound_km)
]

print("Q1:", Q1_km)
print("Q3:", Q3_km)
print("IQR:", IQR_km)
print("Lower Bound:", lower_bound_km)
print("Upper Bound:", upper_bound_km)
print("Number of km-driven outliers:", len(km_outliers))
```

# Conclusion:
# 106 cars were identified as km-driven outliers using the IQR method.
# These observations were retained because high mileage can occur in
# genuine used-car listings and the values are not necessarily erroneous.
